```python
%%writefile app.py

import streamlit as st
from datetime import date

# ============================================================
# EXECUTIVE PRODUCTIVITY AGENT
# Assignment: Arjun Malhotra - VP Sales
# Week: 21 Sep 2026 - 25 Sep 2026
# ============================================================

st.set_page_config(
    page_title="Executive Productivity Agent",
    page_icon="⚡",
    layout="wide"
)

# ============================================================
# DATA
# ============================================================

ACTIONS = [
    {
        "id": "vendor",
        "title": "Send updated vendor list to Raghav",
        "owner": "Arjun Malhotra",
        "type": "My Action",
        "status": "Overdue",
        "deadline": "Wed, 23 Sep 2026 morning",
        "date": "2026-09-23",
        "priority": "High",
        "person": "Raghav Sethi",
        "source": "Leadership Sync + Email + Voice Note",
        "details": (
            "Arjun originally committed to send the updated vendor list "
            "by Tuesday EOD, then moved the commitment to Wednesday morning."
        )
    },
    {
        "id": "campaign",
        "title": "Review Q3 campaign deck",
        "owner": "Arjun Malhotra",
        "type": "My Action",
        "status": "Upcoming",
        "deadline": "Thu, 24 Sep 2026 at 9:30 AM",
        "date": "2026-09-24",
        "priority": "Medium",
        "person": "Neha Kapoor",
        "source": "Email + Calendar",
        "details": (
            "Neha moved the review from Wednesday to Thursday morning. "
            "Deck was ready and attached Thursday at 8:00 AM."
        )
    },
    {
        "id": "meridian",
        "title": "Attend Meridian Logistics client call",
        "owner": "Arjun Malhotra",
        "type": "My Action",
        "status": "Confirmed",
        "deadline": "Wed, 23 Sep 2026 at 3:00 PM",
        "date": "2026-09-23",
        "priority": "High",
        "person": "Priya Nair",
        "source": "Email + Calendar",
        "details": (
            "Arjun proposed Wednesday 3:00 PM, Priya confirmed, "
            "and Arjun reconfirmed the call at 2:00 PM."
        )
    },
    {
        "id": "expense",
        "title": "Review July expense variance report",
        "owner": "Arjun Malhotra",
        "type": "My Action",
        "status": "Ready for Review",
        "deadline": "Before Thu, 24 Sep 2026 board prep",
        "date": "2026-09-24",
        "priority": "Medium",
        "person": "Divya Rao",
        "source": "Email + Calendar",
        "details": (
            "Arjun requested the report by Wednesday evening. "
            "Divya delivered it Wednesday at 6:00 PM."
        )
    },
    {
        "id": "lease",
        "title": "Clarify owner for Mumbai office lease signature",
        "owner": "UNCLEAR",
        "type": "Unclear Ownership",
        "status": "Unowned",
        "deadline": "Fri, 25 Sep 2026 EOD",
        "date": "2026-09-25",
        "priority": "Critical",
        "person": "Facilities / Raghav / Arjun",
        "source": "Email + Leadership Sync",
        "details": (
            "Facilities says an authorized signature is required by Friday. "
            "Raghav asked who owns it. Divya believes Facilities normally "
            "handles it, but no one has confirmed ownership. Arjun explicitly "
            "said: flag it, don't assume."
        )
    }
]

WAITING = [
    {
        "title": "Q3 campaign deck",
        "person": "Neha Kapoor",
        "status": "Delivered",
        "deadline": "Thu, 24 Sep 2026",
        "details": "Deck was ready Thursday at 8:00 AM ahead of the 9:30 AM review."
    },
    {
        "title": "July expense variance report",
        "person": "Divya Rao",
        "status": "Delivered",
        "deadline": "Wed, 23 Sep 2026 evening",
        "details": "Report delivered Wednesday at 6:00 PM."
    }
]

BRIEFS = {
    "2026-09-21": {
        "label": "Monday, 21 Sep 2026",
        "summary": "Leadership Sync captured several commitments and one unresolved ownership issue.",
        "actions": [
            "Send updated vendor list to Raghav",
            "Reconfirm Meridian Logistics call time",
            "Track unclear ownership of Mumbai office lease"
        ]
    },
    "2026-09-22": {
        "label": "Tuesday, 22 Sep 2026",
        "summary": "Vendor list commitment moved to Wednesday morning. Meridian call was scheduled.",
        "actions": [
            "Send updated vendor list to Raghav",
            "Track Mumbai office lease ownership",
            "Prepare for Meridian Logistics call"
        ]
    },
    "2026-09-23": {
        "label": "Wednesday, 23 Sep 2026",
        "summary": "Vendor list is now overdue. Meridian call is confirmed. Expense report arrived.",
        "actions": [
            "Send updated vendor list to Raghav",
            "Attend Meridian Logistics call at 3:00 PM",
            "Review July expense variance before board prep",
            "Track Mumbai lease ownership"
        ]
    },
    "2026-09-24": {
        "label": "Thursday, 24 Sep 2026",
        "summary": "Campaign deck is ready for review. Mumbai lease signature remains unowned and is due tomorrow.",
        "actions": [
            "Review Q3 campaign deck at 9:30 AM",
            "Review July expense variance report",
            "Clarify Mumbai office lease signature owner"
        ]
    },
    "2026-09-25": {
        "label": "Friday, 25 Sep 2026",
        "summary": "Mumbai office lease signature deadline is today and ownership is still unclear.",
        "actions": [
            "Confirm who owns Mumbai office lease signature",
            "Ensure lease signature is completed by EOD"
        ]
    }
}

# ============================================================
# HEADER
# ============================================================

st.title("⚡ Executive Productivity Agent")
st.caption("AI-powered daily action brief for Arjun Malhotra · VP Sales")

st.divider()

# ============================================================
# SIDEBAR
# ============================================================

st.sidebar.header("Executive")

st.sidebar.info(
    "**Arjun Malhotra**\n\n"
    "VP Sales\n\n"
    "veridian-corp.example"
)

st.sidebar.header("Brief Date")

selected_date = st.sidebar.date_input(
    "Select date",
    value=date(2026, 9, 23),
    min_value=date(2026, 9, 21),
    max_value=date(2026, 9, 25)
)

date_key = selected_date.strftime("%Y-%m-%d")

if st.sidebar.button("🔄 Generate Daily Brief", use_container_width=True):
    st.session_state["generated"] = True

st.sidebar.divider()

st.sidebar.caption("Data Pack")
st.sidebar.write("📧 Email threads")
st.sidebar.write("📅 Calendar events")
st.sidebar.write("🎙️ Voice notes")
st.sidebar.write("📝 Meeting notes")

# ============================================================
# DAILY BRIEF
# ============================================================

brief = BRIEFS[date_key]

st.header("📋 Daily Brief")
st.subheader(brief["label"])

st.info(brief["summary"])

# ============================================================
# TOP METRICS
# ============================================================

col1, col2, col3, col4 = st.columns(4)

with col1:
    st.metric("My Actions", "4")

with col2:
    st.metric("Waiting on Others", "2")

with col3:
    st.metric("Overdue", "1", delta="Action required", delta_color="inverse")

with col4:
    st.metric("Unclear Ownership", "1", delta="Needs clarification", delta_color="inverse")

st.divider()

# ============================================================
# TODAY'S ACTIONS
# ============================================================

st.header("🎯 What needs action?")

for action_text in brief["actions"]:
    st.write(f"☐ **{action_text}**")

st.divider()

# ============================================================
# TABS
# ============================================================

tab1, tab2, tab3, tab4, tab5 = st.tabs([
    "🎯 My Actions",
    "⏳ Waiting on Others",
    "🔴 Overdue",
    "⚠️ Unclear Ownership",
    "🔎 Ask the Agent"
])

# ============================================================
# MY ACTIONS
# ============================================================

with tab1:

    st.subheader("Actions owned by Arjun")

    my_actions = [a for a in ACTIONS if a["type"] == "My Action"]

    for action in my_actions:

        if action["status"] == "Overdue":
            icon = "🔴"
        elif action["status"] == "Confirmed":
            icon = "🟢"
        else:
            icon = "🟡"

        with st.expander(f"{icon} {action['title']}"):

            c1, c2, c3 = st.columns(3)

            with c1:
                st.write("**Status**")
                st.write(action["status"])

            with c2:
                st.write("**Deadline**")
                st.write(action["deadline"])

            with c3:
                st.write("**Priority**")
                st.write(action["priority"])

            st.write(f"**Person:** {action['person']}")
            st.write(f"**Source:** {action['source']}")

            st.info(action["details"])

# ============================================================
# WAITING
# ============================================================

with tab2:

    st.subheader("Items received from others")

    for item in WAITING:

        with st.expander(f"📨 {item['title']} — {item['person']}"):

            st.write(f"**Status:** {item['status']}")
            st.write(f"**Expected:** {item['deadline']}")
            st.write(item["details"])

# ============================================================
# OVERDUE
# ============================================================

with tab3:

    st.subheader("🔴 Overdue commitments")

    overdue = [a for a in ACTIONS if a["status"] == "Overdue"]

    if overdue:

        for action in overdue:

            st.error(
                f"**{action['title']}**\n\n"
                f"Deadline: {action['deadline']}\n\n"
                f"Committed to: {action['person']}"
            )

            st.caption(f"Detected from: {action['source']}")

    else:
        st.success("No overdue commitments.")

# ============================================================
# UNCLEAR OWNERSHIP
# ============================================================

with tab4:

    st.subheader("⚠️ Ownership requiring clarification")

    unclear = [
        a for a in ACTIONS
        if a["type"] == "Unclear Ownership"
    ]

    for action in unclear:

        st.warning(
            f"### {action['title']}\n\n"
            f"**Deadline:** {action['deadline']}\n\n"
            f"**Current owner:** {action['owner']}\n\n"
            f"**Potential parties:** {action['person']}"
        )

        st.write(action["details"])

        st.markdown(
            "**Agent decision:** Do not assign ownership automatically."
        )

# ============================================================
# ASK THE AGENT
# ============================================================

with tab5:

    st.subheader("🔎 Ask the Executive Agent")

    question = st.text_input(
        "Ask something about Arjun's commitments",
        placeholder="e.g. What did I promise Raghav?"
    )

    if question:

        q = question.lower()

        # -----------------------------------------------
        # RAGHAV
        # -----------------------------------------------

        if "raghav" in q and (
            "promise" in q or
            "promised" in q or
            "commit" in q or
            "owe" in q or
            "vendor" in q
        ):

            st.success("### You promised Raghav:")

            st.write(
                "**Send the updated vendor list.**"
            )

            st.write(
                "Latest commitment: **Wednesday morning, "
                "23 September 2026.**"
            )

            st.error(
                "⚠️ This commitment is currently overdue."
            )

            st.caption(
                "Sources: Leadership Sync, Vendor List email thread, "
                "and Arjun's voice note."
            )

        # -----------------------------------------------
        # TODAY
        # -----------------------------------------------

        elif "today" in q or "action" in q:

            st.success("### Actions requiring attention:")

            for action in brief["actions"]:
                st.write(f"• {action}")

        # -----------------------------------------------
        # LEASE
        # -----------------------------------------------

        elif "lease" in q or "mumbai" in q:

            st.warning("### Mumbai office lease")

            st.write(
                "The lease signature is due **Friday, 25 September "
                "2026 EOD**."
            )

            st.write(
                "**Ownership is still unclear.**"
            )

            st.write(
                "The agent will not assume Facilities owns it, "
                "because no one has explicitly confirmed ownership."
            )

        # -----------------------------------------------
        # EXPENSE
        # -----------------------------------------------

        elif "expense" in q or "divya" in q:

            st.success("### July expense variance")

            st.write(
                "Divya delivered the report Wednesday at **6:00 PM**."
            )

            st.write(
                "Arjun received it and should review it before "
                "Thursday's board preparation."
            )

        # -----------------------------------------------
        # MERIDIAN
        # -----------------------------------------------

        elif "meridian" in q or "priya" in q or "client" in q:

            st.success("### Meridian Logistics")

            st.write(
                "The client call is **confirmed for Wednesday, "
                "23 September at 3:00 PM**."
            )

            st.write(
                "Arjun proposed the time, Priya confirmed it, "
                "and Arjun reconfirmed the call."
            )

        # -----------------------------------------------
        # CAMPAIGN
        # -----------------------------------------------

        elif "campaign" in q or "deck" in q:

            st.success("### Q3 Campaign Deck")

            st.write(
                "Neha's deck review with Arjun is scheduled for "
                "**Thursday, 24 September at 9:30 AM**."
            )

            st.write(
                "The deck was ready and sent Thursday at 8:00 AM."
            )

        # -----------------------------------------------
        # FALLBACK
        # -----------------------------------------------

        else:

            st.info(
                "I can answer questions about commitments, "
                "deadlines, ownership, overdue items, and waiting items."
            )

            st.write("Try:")

            st.write("• What did I promise Raghav?")
            st.write("• What needs action today?")
            st.write("• What's happening with the Mumbai lease?")
            st.write("• What did Divya send?")
            st.write("• When is the Meridian call?")
            st.write("• When is the campaign deck review?")

# ============================================================
# SOURCE EVIDENCE
# ============================================================

st.divider()

st.header("🔗 Commitment Evidence")

st.caption(
    "The agent deduplicates commitments across multiple sources "
    "and keeps the underlying evidence visible."
)

source_cols = st.columns(4)

with source_cols[0]:
    st.markdown("### 📧 Email")
    st.write("Vendor List")
    st.write("Campaign Deck")
    st.write("Meridian Logistics")
    st.write("Expense Variance")
    st.write("Mumbai Lease")

with source_cols[1]:
    st.markdown("### 📅 Calendar")
    st.write("Leadership Sync")
    st.write("Meridian Call")
    st.write("Board Prep")
    st.write("Facilities Check-in")

with source_cols[2]:
    st.markdown("### 🎙️ Voice")
    st.write("Vendor list reminder")
    st.write("Mumbai lease concern")
    st.write("Expense deadline")

with source_cols[3]:
    st.markdown("### 🧠 Agent Logic")
    st.write("✓ Deduplicates")
    st.write("✓ Detects deadlines")
    st.write("✓ Detects overdue items")
    st.write("✓ Separates ownership")
    st.write("✓ Flags ambiguity")

# ============================================================
# FOOTER
# ============================================================

st.divider()

st.caption(
    "Executive Productivity Agent · Prototype · "
    "Veridian Corp · Week of 21–25 September 2026"
)
```
